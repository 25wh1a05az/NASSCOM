```python
# Core imports for the whole lab
import numpy as np
import sympy as sp

x, y = sp.symbols('x y')      # symbolic variables we'll reuse
sp.init_printing()            # pretty-print symbolic math
np.random.seed(42)
print('Setup complete. SymPy', sp.__version__, '| NumPy', np.__version__)
```

```python


# -----------------------------------------------------------
# 🔹 1A. NUMERICAL DERIVATIVE (finite difference)
# -----------------------------------------------------------

# The derivative is the slope: how much f changes for a tiny step h
def f(x):
    return x**2

h = 1e-6
slope_at_3 = (f(3 + h) - f(3)) / h
print('Numerical f\'(3):', round(slope_at_3, 4), ' (exact = 6)')


# -----------------------------------------------------------
# 🔹 1B. SYMBOLIC DERIVATIVE (SymPy)
# -----------------------------------------------------------

expr = x**2
deriv = sp.diff(expr, x)          # differentiate w.r.t. x
print('d/dx (x**2) =', deriv)     # 2*x

# Evaluate the symbolic derivative at x = 3
print('Symbolic f\'(3) =', deriv.subs(x, 3))

```

```python
import sympy as sp

def g(x):
    return x**3 + 2*x

# 1. Numerical derivative at x = 2 using finite difference
x = 2
h = 1e-6

numerical_derivative = (g(x + h) - g(x)) / h

print("Numerical derivative:", numerical_derivative)


# 2. Symbolic derivative
x_sym = sp.symbols('x')

g_sym = x_sym**3 + 2*x_sym
symbolic_derivative = sp.diff(g_sym, x_sym)

print("Symbolic derivative:", symbolic_derivative)


# 3. Evaluate symbolic derivative at x = 2 and compare
symbolic_value = symbolic_derivative.subs(x_sym, 2)

print("Symbolic derivative at x = 2:", symbolic_value)
print("They match:", abs(numerical_derivative - float(symbolic_value)) < 1e-5)
```

```python
import sympy as sp

# Create fresh symbolic variables
x, y = sp.symbols('x y')

# -----------------------------------------------------------
# 🔹 2A. PARTIAL DERIVATIVES
# -----------------------------------------------------------

f2 = x**2 + 3*x*y + y**2

# Partial derivatives
print('df/dx =', sp.diff(f2, x))
print('df/dy =', sp.diff(f2, y))


# -----------------------------------------------------------
# 🔹 2B. THE GRADIENT
# -----------------------------------------------------------

# Gradient = vector of partial derivatives
grad = [sp.diff(f2, v) for v in (x, y)]

print('grad f =', grad)


# Evaluate gradient at (x=1, y=2)
grad_at = [g.subs({x: 1, y: 2}) for g in grad]

print('grad f at (1, 2) =', grad_at)
```

```python
import sympy as sp

# Create symbolic variables
x, y = sp.symbols('x y')

h2 = x**2 * y + sp.sin(y)

# 1. dh/dx and dh/dy
dh_dx = sp.diff(h2, x)
dh_dy = sp.diff(h2, y)

print("dh/dx =", dh_dx)
print("dh/dy =", dh_dy)

# 2. Assemble the gradient
grad_h = [dh_dx, dh_dy]

print("Gradient =", grad_h)

# 3. Evaluate gradient at (x=2, y=0)
grad_at = [g.subs({x: 2, y: 0}) for g in grad_h]

print("Gradient at (2, 0) =", grad_at)
```

```python


# -----------------------------------------------------------
# 🔹 3A. CHAIN RULE BY HAND vs SymPy
# -----------------------------------------------------------

# y = sin(x**2) is a composition: outer = sin(u), inner = u = x**2
# Chain rule:  dy/dx = cos(u) * du/dx = cos(x**2) * 2x
by_hand = sp.cos(x**2) * 2*x
by_sympy = sp.diff(sp.sin(x**2), x)

print('By hand :', by_hand)
print('By SymPy:', by_sympy)
print('Match?  ', sp.simplify(by_hand - by_sympy) == 0)


# -----------------------------------------------------------
# 🔹 3B. CHAINING THREE FUNCTIONS
# -----------------------------------------------------------

# y = (3x + 1)**4  -> outer^4, inner (3x+1)
expr3 = (3*x + 1)**4
print('d/dx (3x+1)^4 =', sp.diff(expr3, x))   # 12*(3x+1)^3

```

```python
import sympy as sp

# Create symbolic variable
x = sp.symbols('x')

# Function
y = sp.exp(x**2 + 1)

# 1. Chain rule by hand:
# outer derivative = exp(x**2 + 1)
# inner derivative = 2*x
by_hand = sp.exp(x**2 + 1) * (2*x)

print("By hand:", by_hand)

# 2. Differentiate using SymPy
by_sympy = sp.diff(y, x)

print("With SymPy:", by_sympy)

# 3. Confirm they match
print("They match:", sp.simplify(by_hand - by_sympy) == 0)
```

```python


# -----------------------------------------------------------
# 🔹 4A. FORWARD PASS
# -----------------------------------------------------------

# Tiny toy problem: 4 samples, 3 input features, 5 hidden units, 1 output
X = np.random.randn(4, 3)
Y = np.random.randn(4, 1)
W1 = np.random.randn(3, 5) * 0.1
W2 = np.random.randn(5, 1) * 0.1

z1 = X @ W1                 # linear layer 1
h  = np.maximum(0, z1)      # ReLU activation
y_hat = h @ W2              # linear layer 2 (prediction)
loss = ((y_hat - Y) ** 2).mean()
print('Initial loss:', round(loss, 4))


# -----------------------------------------------------------
# 🔹 4B. BACKWARD PASS (gradients via the chain rule)
# -----------------------------------------------------------

# Work backwards from the loss, one link at a time
dy   = 2 * (y_hat - Y) / Y.size      # d loss / d y_hat
dW2  = h.T @ dy                      # d loss / d W2
dh   = dy @ W2.T                     # d loss / d h
dz1  = dh * (z1 > 0)                 # ReLU gradient (1 where z1>0 else 0)
dW1  = X.T @ dz1                     # d loss / d W1

print('dW1 shape:', dW1.shape, '(matches W1)')
print('dW2 shape:', dW2.shape, '(matches W2)')


# -----------------------------------------------------------
# 🔹 4C. ONE GRADIENT-DESCENT STEP SHOULD LOWER THE LOSS
# -----------------------------------------------------------

lr = 0.1
W1 -= lr * dW1               # step downhill
W2 -= lr * dW2

# Recompute the loss after the update
h_new = np.maximum(0, X @ W1)
loss_new = ((h_new @ W2 - Y) ** 2).mean()
print('Loss before:', round(loss, 4))
print('Loss after :', round(loss_new, 4), '-> should be lower')

```

```python
import numpy as np

# Fresh weights/data
Xb = np.random.randn(6, 4)
Yb = np.random.randn(6, 1)

Wa = np.random.randn(4, 8) * 0.1     # input -> hidden
Wb = np.random.randn(8, 1) * 0.1     # hidden -> output


# ReLU function
def relu(x):
    return np.maximum(0, x)


# 1. FORWARD PASS

z1 = Xb @ Wa
h = relu(z1)

y_hat = h @ Wb

# Mean Squared Error loss
loss = np.mean((y_hat - Yb) ** 2)

print("Loss before:", loss)


# 2. BACKWARD PASS

# dL/dy_hat
dy = (2 / Yb.shape[0]) * (y_hat - Yb)

# dL/dWb
dWb = h.T @ dy

# dL/dh
dh = dy @ Wb.T

# ReLU derivative
dz1 = dh * (z1 > 0)

# dL/dWa
dWa = Xb.T @ dz1


# 3. ONE GRADIENT-DESCENT STEP

lr = 0.05

Wa = Wa - lr * dWa
Wb = Wb - lr * dWb


# Forward pass again to calculate new loss
z1_new = Xb @ Wa
h_new = relu(z1_new)
y_hat_new = h_new @ Wb

loss_after = np.mean((y_hat_new - Yb) ** 2)

print("Loss after:", loss_after)
```

```python
import sympy as sp

# Reset symbolic variables
x, y = sp.symbols('x y')

# -----------------------------------------------------------
# 🔹 5A. THE HESSIAN
# -----------------------------------------------------------

f5 = x**2 + 3*x*y + y**2

H = sp.hessian(f5, (x, y))

print('Hessian of f:')
sp.pprint(H)


# -----------------------------------------------------------
# 🔹 5B. GRADIENT DESCENT ON A SIMPLE FUNCTION
# -----------------------------------------------------------

# Minimise f(x) = (x - 4)**2
# Derivative: f'(x) = 2*(x - 4)

xv = 0.0
lr = 0.2

for step in range(15):
    grad = 2 * (xv - 4)
    xv = xv - lr * grad

print('Converged x:', round(xv, 3), ' (true minimum = 4)')
```

```python
import sympy as sp

# Reset symbolic variables
x, y = sp.symbols('x y')


# 1. Hessian of f = x**4 + y**2

f = x**4 + y**2

H = sp.hessian(f, (x, y))

print("Hessian:")
sp.pprint(H)


# 2. Gradient descent to minimise (x - 7)**2

xv = 0.0
lr = 0.1

for step in range(20):
    grad = 2 * (xv - 7)
    xv = xv - lr * grad

print("\nFinal x:", xv)
```
